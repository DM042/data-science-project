```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/heart.csv")
print(df.columns.tolist())
print(df.head())
print(df.shape)
```

```python
pain_rate = df.groupby("ChestPainType")["HeartDisease"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
pain_rate.plot(kind="bar", color="#c0392b")
plt.ylabel("Heart Disease Rate")
plt.xlabel("Chest Pain Type")
plt.title("Heart Disease Rate by Chest Pain Type")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("fig1_chest_pain.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(x="HeartDisease", y="MaxHR", data=df, palette="Set2")
plt.xticks([0, 1], ["No Disease", "Heart Disease"])
plt.ylabel("Maximum Heart Rate")
plt.xlabel("")
plt.title("MaxHR by Heart Disease Status")
plt.tight_layout()
plt.savefig("fig2_maxhr_boxplot.png", dpi=150)
plt.show()
```
